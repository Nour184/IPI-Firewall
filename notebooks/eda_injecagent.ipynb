```python
import json
from pathlib import Path
import pandas as pd
from IPython.display import display

INJEC_DIR = Path("..") / "dataset" / "raw" / "injecagent"

def load_any_json(path: Path):
    text = path.read_text(encoding="utf-8").strip()
    if path.suffix == ".jsonl":
        return [json.loads(line) for line in text.splitlines() if line.strip()]
    return json.loads(text)

summary_rows = []
loaded_files = {}

for path in sorted(INJEC_DIR.rglob("*")):
    if path.suffix not in {".json", ".jsonl"}:
        continue
    data = load_any_json(path)
    loaded_files[path.name] = data

    sample = data[0] if isinstance(data, list) and data else (next(iter(data.values())) if isinstance(data, dict) and data else {})
    keys = list(sample.keys()) if isinstance(sample, dict) else [type(sample).__name__]

    summary_rows.append({
        "file": str(path.relative_to(INJEC_DIR)),
        "container": type(data).__name__,
        "rows": len(data),
        "keys": ", ".join(keys),
    })

pd.set_option("display.max_colwidth", None)
print("1. InjecAgent Files & Schema Overview")
display(pd.DataFrame(summary_rows))

# Preview the first row of the first test_cases file
first_tc = next((name for name in loaded_files if "test_cases" in name), next(iter(loaded_files)))
print(f"\n2. Sample Record from: {first_tc}")
print(json.dumps(loaded_files[first_tc][0], indent=2, ensure_ascii=False))
```
